```python
# 1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([
[1,2],
[3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of Ones:")
print(ones_array)

print ("Array of zeros:")
print(zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
array_6x3 = np.random.rand(6, 3)
transposed_array = array_6x3.T

print("Original 6x3 array:")
print(array_6x3)

print("Transposed array:")
print(transposed_array)
```

```python
# 5. Create two random numpy arrays of size n x p and p x m and compute the dot product
n, p, m = 3, 4, 2

array_np = np.random.rand(n, p)
array_pm = np.random.rand(p, m)
dot_product = np.dot(array_np, array_pm)

print("Array n x p:")
print(array_np)

print("Array p x m:")
print(array_pm)

print("Dot product:")
print(dot_product)
```
